```python
import numpy as np

from typing import Dict, Tuple
```

```python


class FederatedClient:
    def __init__(self, data: np.ndarray, labels: np.ndarray, **kwargs):
        """
        Initialize a federated learning client
        
        Args:
            data: Local training data (features)
            labels: Corresponding labels
            **kwargs: Client configuration parameters that can override defaults
        """
        self.data = data
        self.labels = labels
        
        
        self.connection_quality = kwargs.get('connection_quality', 
                                           np.random.choice(['poor', 'fair', 'good', 'excellent']))
        self.battery = kwargs.get('battery', np.random.uniform(0.2, 1.0))  
        self.computation_power = kwargs.get('computation_power', 
                                          np.random.choice(['low', 'medium', 'high']))
        
        
        self.model # = <insert model>
        if data is not None:
            self.model.fit(data, labels)  

    def get_client_info(self) -> Dict:
        """client data for server selection"""
        return {
            'dataset_size': len(self.data),
            'connection_quality': self.connection_quality,
            'battery': self.battery,
            'computation_power': self.computation_power
        }

    def train_local_model(self, global_params: np.ndarray) -> Tuple[np.ndarray, Dict]:
        """
        Train local model using current global parameters
        
        Args:
            global_params: Model parameters from the global model
            
        Returns:
            Tuple of (updated parameters, training metadata)
        """
        
        self.model.coef_ = global_params
        
        
        self.model.fit(self.data, self.labels)
        
        
        loss = self.model.score(self.data, self.labels)
        
        return self.model.coef_, {
            'loss': loss,
            'samples_used': len(self.data),
            'client_info': self.get_client_info()
        }
```
